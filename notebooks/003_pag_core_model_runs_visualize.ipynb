```python
import pandas as pd
import numpy as np

# Extract the model names and R2 values
model_names = mm.function.regression.estimators_[0].r2.index
r2_values = mm.function.regression.estimators_[0].r2.values

coeffs = mm.function.regression.estimators_[0].coef_
t_stats = mm.function.regression.estimators_[0].t_
intercepts = mm.function.regression.estimators_[0].intercept_

# Create a DataFrame with the model names and R2 values
df = pd.DataFrame({'model_name': model_names, 'r2': r2_values})

# Create a MultiIndex DataFrame with the coefficients and t-statistics
coeff_df = pd.DataFrame({'coeff_value': coeffs.flatten(), 't_stat': t_stats.flatten()})
coeff_df['coeff_name'] = ['coeff_' + str(i).zfill(2) for i in range(coeffs.shape[1])] * coeffs.shape[0]

# Add the intercepts to the coefficient DataFrame
intercept_df = pd.DataFrame({'coeff_value': intercepts, 't_stat': np.nan, 'coeff_name': ['intercept'] * len(intercepts)})

# Concatenate the coefficient DataFrame and the intercept DataFrame
coeff_df = pd.concat([coeff_df, intercept_df])

# Rearrange the columns to match the desired output
coeff_df = coeff_df[['coeff_name', 'coeff_value', 't_stat']]

# Add the model names to the DataFrame
coeff_df['model_name'] = np.repeat(model_names, coeffs.shape[1] + 1)

# Add the R2 values to the DataFrame
coeff_df['model_r2'] = np.repeat(r2_values, coeffs.shape[1] + 1)

coeff_df
```
