```python
scope = emat.Scope('EMAT/PAG/PAG_EMAT_scope_final.yml')
```

```python
scope.info()
```

```python
# Read Experiments
experiments = pd.read_csv(r"EMAT/PAG/pag_emat_design_experiments_v2.csv")
```

```python
experiments
```

```python
results = pd.read_csv(os.path.join(model_results_dir, "final_merged_results.csv"))
```

```python
results['experiment'] = results['RunID'].str.extract('(\d+)').astype(int)
results = results.drop(columns=['RunID'])
```

```python
results.shape
```

```python
experiments_df = pd.merge(experiments, results, on='experiment', how='inner')
experiments_df
```

```python
cat_perf_measures = ['Roadway_Capacity_Improvement', 'Transit_Service_Expansion']

#convert to categrorical data type
for col in cat_perf_measures:
    experiments_df[col] = pd.Categorical(experiments_df[col], ordered=True)
```

```python
# save results as parquet files and csv
#experiments_df.to_parquet("EMAT\\PAG\\Core Model.pq")
#experiments_df.to_csv("EMAT\\PAG\\Core Model.csv", index=False)
```

```python
# FinalPriority
perf_df = pd.read_excel(
    os.path.join("crosswalks", "perf_measures_names.xlsx")
)

perf_df = perf_df.loc[perf_df['FinalPriority'].isin(['Y1', 'Y2'])]
perf_df = list(perf_df['PF Name in the Dashboard'])

measures_for_meta_models = perf_df
```

```python
measures_for_meta_models
```

```python
%%time 

# Create MetaModels
mm = emat.create_metamodel(scope, experiments_df, suppress_converge_warnings=False, include_measures=measures_for_meta_models)
```

```python
# extract R2 values from Linear Regression
mm.function.regression.estimators_[0].r2
```

```python
# Extract Coeff from Linear Regression
mm.function.regression.estimators_[0].coef_
```

```python
# Extract Intercepts from Linear Regression
mm.function.regression.estimators_[0].intercept_
```

```python
# Extract p values from Linear Regression
mm.function.regression.estimators_[0].p_
```

```python
# Extract t values from Linear Regression
mm.function.regression.estimators_[0].t_
```

```python
design = mm.design_experiments(n_samples=10000, sampler='lhs')
mm_result = mm.run_experiments(design)
mm_result = mm_result.reset_index().rename(columns={'index':'experiment'}).set_index('experiment')


from emat.analysis import Visualizer
viz = Visualizer(scope=scope, data=mm_result)
```

```python
with timing_log("crossvalidate metamodel"):
    display(mm.cross_val_scores())
```
