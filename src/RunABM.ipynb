```python
def replace_text_in_file(input_file_path, search_text, replace_text, output_file_path):
    try:
        # Read the contents of the input file
        with open(input_file_path, 'r') as file:
            content = file.read()

        # Replace the target text
        new_content = content.replace(search_text, replace_text)

        # Write the new content to the output file
        with open(output_file_path, 'w') as file:
            file.write(new_content)

        print(f"Text replacement completed successfully. Saved to '{output_file_path}'.")

    except FileNotFoundError:
        print(f"Error: The file '{input_file_path}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# Usage
#if __name__ == "__main__":
    # Specify the input file path, text to search for, replacement text, and output file path
input_file_path = 'phoenix_ui_template.lst'  # Change this to your input file path
search_text = 'PAGABMPATH'        # Text to search for
replace_text = model_path      # Text to replace with
output_file_path =  os.path.join(model_path,"gisdk",'phoenix_ui.lst') # Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)

input_file_path = 'phoenix20170523_template.ini'  # Change this to your input file path
search_text = 'PAGABMPATH'        # Text to search for
replace_text = model_path     # Text to replace with
output_file_path = os.path.join(model_path,'phoenix20170523.ini') # Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)

input_file_path = '20190227_15_18_16b_16b_PopSyn3_CTRAMP2_TC60_template.rsc'  # Change this to your input file path
search_text = 'PAGABMPATH'        # Text to search for
replace_text = model_path_double     # Text to replace with
output_file_path = os.path.join(model_path,"gisdk",'20190227_15_18_16b_16b_PopSyn3_CTRAMP2_TC60.rsc') # Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)

input_file_path = 'mag_abm_template.properties'  # Change this to your input file path
search_text = 'PAGABMPATH'        # Text to search for
replace_text = model_path_slash     # Text to replace with
output_file_path = os.path.join(model_path,"runtime","config",'mag_abm.properties') # Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)

input_file_path = 'PerformanceMeasure_ABM_1.rsc'  # Change this to your input file path
search_text = 'MODELFOLDER'        # Text to search for
replace_text = model_path_double + "\\\\"     # Text to replace with
output_file_path = os.path.join(model_path,"gisdk",'PerformanceMeasure_ABM.rsc')# Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)
search_text = 'SCENARIONAMEREPLACE'
replace_text = scenario_name
replace_text_in_file(output_file_path, search_text, replace_text, output_file_path)

input_file_path = "SimplePerformanceMeasuresTemplate.R"
search_text = "MODELFOLDERREPLACE"
replace_text = model_path_double     # Text to replace with
output_file_path = "SimplePerformanceMeasuresRun.R"# Change this to your desired output file path
replace_text_in_file(input_file_path, search_text, replace_text, output_file_path)
search_text = 'SCENARIONAMEREPLACE'
replace_text = scenario_name
replace_text_in_file(output_file_path, search_text, replace_text, output_file_path)
```

```python
def update_argument_in_gisdk_file(file_path, arg, value):
    # Read the file
    with open(file_path, 'r') as file:
        lines = file.readlines()

    # Iterate through each line
    for i, line in enumerate(lines):
        # Check if the line starts with the specific string followed by '='
        if arg + '=' in line:
            lines[i] = '\t' + arg + '=' + "'" + value + "'" + '\n'

    # Write back to the file
    with open(file_path, 'w') as file:
        file.writelines(lines)
```

```python
gisdk_file = os.path.join(config['src_dir'], "process_model_outputs.rsc")

update_argument_in_gisdk_file(gisdk_file, "model_folder", config['abm_model_path'].replace('/','\\\\'))
update_argument_in_gisdk_file(gisdk_file, "output_folder", os.path.join(abm_path,"out").replace('/','\\\\'))
update_argument_in_gisdk_file(gisdk_file, "iteration", "3")
```
